### 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(between) · 1·2등석(isin) 조건으로 pclass, age, fare, survived 열을 선택하세요.

```python
titanic01 = titanic[(titanic["sex"] == "male") &
                    (titanic["age"].between(20,40)) & 
                    (titanic["pclass"].isin([1,2]))]
titanic01 = titanic01[["pclass", "age", "fare", "survived"]]
titanic01
```

2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.

```python
# 행 수
titanic01.shape[0]
```

```python
# 생존률
round(titanic01["survived"].mean(),2)
```

```python
#요금 상위 3명
titanic01.sort_values("fare", ascending=False).head(3)
```

3. 같은 조건을 query로 써서 행 수를 검증하세요.

```python
# 남성 · 나이 20~40세(between) · 1·2등석(isin) 
titanic02 = titanic.query("sex == 'male' and age >= 20 and age <= 40 and pclass in [1,2]")
titanic02.shape[0]
```

4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
survived_12 = round(titanic01["survived"].mean(),2)
titanic02 = titanic[(titanic["sex"] == "male") &
                    (titanic["age"].between(20,40)) & 
                    (titanic["pclass"].isin([3]))]["survived"]
survived_3 = round(titanic02.mean(),2)

print(survived_12, survived_3)
```

20~40세 남성 승객의 객실 등급별 평균 생존률을 비교한 결과, 1·2등급 승객의 생존률은 24%로 

3등급 승객의 생존률인 15%보다 9%p 높게 나타났다. 

다만 연령과 성별 외의 다른 요인들을 통제하지 않았으므로, 객실 등급이 높을수록 생존률이 높았다고 

단정할 수 없다.
