### 1. 파일 읽기

- `baemin` 폴더의 파일 31개를 모두 읽어서 하나의 DataFrame으로 만드세요.
- `coupang` 폴더의 파일 31개도 같은 방식으로 읽으세요.
- 파일 목록은 아래 힌트 코드로 가져오고, 반복문으로 하나씩 읽어서 합치세요.

```python
# 배민 dataframe 31개 합치기
baemin_files = sorted(glob.glob("data/baemin/baemin*.csv"))

df_baemin_list = []
for f in baemin_files :
    df = pd.read_csv(f)
    df_baemin_list.append(df)
    
df_baemin = pd.concat(df_baemin_list, ignore_index=True)
df_baemin
```

```python
# 쿠팡 dataframe 31개 합치기
coupang_files = sorted(glob.glob("data/coupang/coupang*.csv"))

df_coupang_list = []
for f in coupang_files :
    df = pd.read_csv(f)
    df_coupang_list.append(df)
    
df_coupang = pd.concat(df_coupang_list, ignore_index = True)
df_coupang
```

### 2. 칼럼명 통일

- 두 플랫폼의 칼럼명을 `주문일시`, `전체금액`, `광고유입여부`로 바꾸세요.

```python
# 배민["order_at", "total_bill", "ad_yn", "cpc"]
df_baemin = df_baemin.rename(columns={
    "order_at"   : "주문일시",
    "total_bill" : "전체금액",
    "ad_yn"      : "광고유입여부",   
})
print("배민 :",df_baemin)

# 쿠팡["order_time", "total_price", "advertisement_yn", "ad_pct"
df_coupang = df_coupang.rename(columns={
    "order_time"        : "주문일시",
    "total_price"       : "전체금액",
    "advertisement_yn"  : "광고유입여부",   
})
print("쿠팡 :", df_coupang)
```

- 어느 플랫폼 데이터인지 알 수 있도록 `플랫폼` 칼럼(`배민` / `쿠팡`)을 추가하세요.

```python
df_baemin["플랫폼"]="배민"
print("배민 :",df_baemin)

df_coupang["플랫폼"]="쿠팡"
print("쿠팡 :", df_coupang)
```

- 두 플랫폼 모두 주문 1건마다 `광고집행비`(원) 칼럼을 만드세요.
    - 배민: `cpc` 그대로
    - 쿠팡: `전체금액 × ad_pct / 100`

```python
df_baemin["광고집행비(원)"] = df_baemin["cpc"]
print(df_baemin)


df_coupang["광고집행비(원)"] = (df_coupang["전체금액"]*df_coupang["ad_pct"])/100
print(df_coupang)
```

### 3. 데이터 통합

- 두 DataFrame을 하나로 합치세요. (`pd.concat`)

```python
df_total = pd.concat([df_baemin, df_coupang]
                         , ignore_index = True)

df_total.shape
print(df_total)
```

- `주문일시`를 datetime으로 바꾸고, 날짜만 담은 `주문일자` 칼럼을 추가하세요.

```python
df_total["주문일시"] = pd.to_datetime(df_total["주문일시"])
print(df_total["주문일시"].dtype)

df_total["주문일자"] = df_total["주문일시"].dt.date
```

- 주문일시 순으로 정렬하세요.

```python
df_total = df_total.sort_values("주문일시")
df_total
```

### 4. 일자별 매출 집계

- 행은 `주문일자`, 열은 `플랫폼`, 값은 `전체금액` 합계인 표를 만드세요. (`pivot_table`)

```python
df_pivot = df_total.pivot_table(
    index = "주문일자",
    columns = "플랫폼" ,
    values= "전체금액",
    aggfunc = "sum"
)
df_pivot.head()
```

- `합계` 열을 추가하세요.

```python
df_pivot["합계"] = df_pivot["배민"]+df_pivot["쿠팡"]
df_pivot.head()
```

### 5. 광고 효과 분석

- 플랫폼별로 광고유입 주문이 전체 주문 건수에서 차지하는 **비율(%)**을 구하세요.

```python
a = df_total[df_total["광고유입여부"]=="Y"].groupby(["플랫폼"]).size()
print("플랫폼별 광고유입주문건수:",a)
b= df_total.groupby(["플랫폼"]).size()
print("플랫폼별 전체주문건수:",b)
print("전체 주문건수 대비 광고유입주문건수 비율 :", round((a/b)*100,2))
```

```python
df_bank_trans = pd.read_csv("data/bank_20261001_20261031.csv",
                            encoding = "cp949")

df_bank_trans
bank_trans_sikjaje = df_bank_trans[df_bank_trans["거래처"]=="식자재마트"]
mask = bank_trans_sikjaje["출금액"]!=0
bank_trans_sikjaje = bank_trans_sikjaje[mask]
bank_trans_sikjaje.head()
```

- 일자별 식자재비 합계를 구하세요.

```python
bank_trans_sikjaje["거래일자"] = pd.to_datetime(
    bank_trans_sikjaje["거래일시"]
    ).dt.date
```

```python
total_sikjaje_day = bank_trans_sikjaje.groupby("거래일자")["출금액"].sum()
total_sikjaje_day.head()
```

```python
total_sikjaje_day = total_sikjaje_day.reset_index()
```

### 7. 최종 결과표 저장

- 4번 결과와 6번 결과를 일자 기준으로 합쳐서 아래 칼럼을 가진 표를 만드세요.
- 주문일자	배민	쿠팡	합계	식자재비	매출-식자재비

```python
total_sikjaje_day = total_sikjaje_day.rename(
    columns={
    "거래일자": "주문일자",
    "출금액": "식자재비"
})
```

```python
df_result = pd.merge(
    df_pivot,
    total_sikjaje_day,
    on="주문일자",
    how="left"
)

df_result["식자재비"] = df_result["식자재비"].fillna(0)

df_result["매출-식자재비"] = df_result["합계"] - df_result["식자재비"]

df_result = df_result[
    ["주문일자", "배민", "쿠팡", "합계", "식자재비", "매출-식자재비"]
]


df_result.to_csv("result.csv", index=False, encoding="utf-8-sig")
df_result.head(10)
```
